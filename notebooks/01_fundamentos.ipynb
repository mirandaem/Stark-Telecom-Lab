## Marco teórico

Un sistema de comunicación digital transmite información desde una fuente hacia un destino a través de un canal. En su forma básica está compuesto por transmisor, canal y receptor. La información se representa mediante bits. El canal puede introducir ruido, lo que altera la señal recibida.

Modelo básico del canal:

\[
r = s + n
\]

donde \(r\) es la señal recibida, \(s\) la señal transmitida y \(n\) el ruido.

La tasa de error de bit se define como:

\[
BER = \frac{N_e}{N_t}
\]

donde \(N_e\) es el número de bits erróneos y \(N_t\) el número total de bits transmitidos.

```python
import random
#Función de Transmisión
def transmitir_bits(bits, prob_error=0.1, semilla=None):
    """
    Simula la transmisión de una secuencia binaria a través de un canal simple.
    Cada bit puede invertirse con probabilidad prob_error.
    """
    if semilla is not None:
        random.seed(semilla)

    recibido = ""
    errores = 0

    for bit in bits:
        if random.random() < prob_error:
            recibido += "1" if bit == "0" else "0"
            errores += 1
        else:
            recibido += bit

    return recibido, errores
```

```python
#Función BER
def calcular_ber(original, recibido):
    if len(original) != len(recibido):
        raise ValueError("Las secuencias deben tener la misma longitud.")

    errores = sum(1 for a, b in zip(original, recibido) if a != b)
    ber = errores / len(original)
    return errores, ber
```

```python
#Prueba guiada
mensaje = "1011001"
recibido, errores_canal = transmitir_bits(mensaje, prob_error=0.2, semilla=42)
errores, ber = calcular_ber(mensaje, recibido)

print("Mensaje original :", mensaje)
print("Mensaje recibido :", recibido)
print("Errores detectados:", errores)
print("BER:", ber)
```

```python
#Comparación Bit a Bit
def comparar_bits(original, recibido):
    print("Pos | Tx | Rx | Estado")
    print("-" * 24)
    for i, (a, b) in enumerate(zip(original, recibido), start=1):
        estado = "OK" if a == b else "ERROR"
        print(f"{i:>3} | {a}  | {b}  | {estado}")

comparar_bits(mensaje, recibido)
```
